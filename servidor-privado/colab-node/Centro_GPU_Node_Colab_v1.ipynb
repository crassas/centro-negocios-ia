```python
# 3) Descarregar motor + ponte da Estação
import urllib.request, importlib.util
BASE='https://raw.githubusercontent.com/crassas/centro-negocios-ia/main/servidor-privado/colab-node/'
urllib.request.urlretrieve(BASE+'centro_colab_gpu_node_v1.py','/content/centro_colab_gpu_node_v1.py')
urllib.request.urlretrieve(BASE+'centro_colab_bridge_v1.py','/content/centro_colab_bridge_v1.py')
print('Scripts descarregados.')
```

```python
# 4) Carregar o maior modelo que couber
spec=importlib.util.spec_from_file_location('centro_gpu','/content/centro_colab_gpu_node_v1.py')
centro_gpu=importlib.util.module_from_spec(spec); spec.loader.exec_module(centro_gpu)
ENGINE=centro_gpu.boot()
```

```python
# 5) Teste local antes de ligar ao Telegram
print(ENGINE.chat('Responde apenas: GPU Colab operacional.'))
```

```python
# 6) Ligar à Estação Centro / Telegram
spec2=importlib.util.spec_from_file_location('centro_bridge','/content/centro_colab_bridge_v1.py')
centro_bridge=importlib.util.module_from_spec(spec2); spec2.loader.exec_module(centro_bridge)
centro_bridge.run(ENGINE)
```
